```python
import re
import pandas as pd
import nltk
import spacy
from spacy import displacy

try:
    nltk.data.find("taggers/averaged_perceptron_tagger_eng")
except LookupError:
    nltk.download("averaged_perceptron_tagger_eng")
```

```python
nlp = spacy.load("en_core_web_sm")
```

```python
sentence = "When I arrived, the dog in the yard barked."
words = re.findall(r'\w+|[^\s\w]+', sentence)
print(words)
```

```python
pos_tags = nltk.pos_tag([token for token in words if token not in [",", "."]])
pd.DataFrame(pos_tags).T
```

```python
spacy_pos_tagged = [(word, word.tag_, word.pos_) for word in nlp(sentence)]
print(f"\n{pd.DataFrame(spacy_pos_tagged).T}")
```

```python
displacy.render(nlp(sentence), jupyter=True, options={
    "distance": 100,
    "arrow_stroke": 1.5,
    "arrow_width": 8
})
```

```python
nlp = spacy.load("es_core_news_md")
sentence = nlp("Cuando llegué el perro que estaba en el patio ladró")
displacy.render(nlp(sentence), jupyter=True, options={
    "distance": 100,
    "arrow_stroke": 1.5,
    "arrow_width": 8
})
```

```python
import stanza
stanza.download('es', processors='tokenize,pos,constituency')

nlp = stanza.Pipeline('es', processors='tokenize,pos,constituency')
doc = nlp("Cuando llegué, el perro que estaba en el patio ladró.")

for sent in doc.sentences:
    print(sent.constituency)
```
