```python
import nltk

with open("state_of_the_union_2nd_term.txt", "r", encoding="utf8") as f:
    raw_text = f.read()

tokens = nltk.word_tokenize(raw_text)
text_object = nltk.Text(tokens, name="State of the Union - 2nd Term")
```

```python
text_object.concordance("border")
```

```python
text_object.similar("nation")
```

```python
text_object.common_contexts(["america", "america"])
```

```python
text_object.dispersion_plot(["security", "health", "corruption", "illegal", "terrorist", "love"])
```

```python
text_object.generate()
```

```python
len(text_object)
```

```python
sorted(nltk.FreqDist(text_object).items(), key=lambda item: item[1])
```

```python
def lexical_diversity(text) -> float:
    return len(text) / len(set(text))

def percentage(count, total):
    return 100 * count / total
```

```python
nltk.FreqDist(text_object).plot(10, cumulative=True)
```

```python
nltk.FreqDist(text_object).hapaxes()
```
