````python
from bs4 import BeautifulSoup

path = "./corpora/excelsior/"
filename = "e990505_mod.htm"

with open(path + filename, "r", encoding="utf-8") as f:
    text = f.read()

soup = BeautifulSoup(text, "html.parser")
text = soup.get_text(separator="\n", strip=True)
print(text)
````

### Pre-procesamiento

Cargar la lista de stopwords en español de ```nltk.corpus```

````python
import re
from nltk.corpus import stopwords

stop_es = set(stopwords.words("spanish"))
print(stop_es)
````

Pasar el texto a minúsculas y tokenizarlo. Descartartar:
- El nombre del archivo
- Signos de puntuación
- Correos electrónicos
- Encabezados
- Navegación

````python
text = re.sub(r"\S+\.html?\b", " ", text) # URLs y el nombre del archivo
text = re.sub(r"\S+(@|\(a\))\S+", " ", text)  # correos electrónicos
text = re.sub(r"Excelsior:.*", " ", text) # encabezados
text = re.sub(r"Nota siguiente", " ", text) # navegación
text = re.findall(r"\w+", text.lower())
print(text[:30])
````

Filtrar stopwords y tokens numéricos y alfanuméricos.

````python
text = [token for token in text if token not in stop_es and token.isalpha()]
print(text)
````

````python
with open("tokens.txt", "w+") as file:
    file.write(" ".join(text))
````

Vocabulario y conteo de palabras.

````python
from collections import Counter

word_count = Counter(text)
print(word_count)
````

````python
lexicon = word_count.keys()
lexicon = list(lexicon)
print(lexicon)
````
